Лаба 1. 
Вычислить сумму s, прекращая суммирование, когда очередной член суммы
по абсолютной величине станет меньше 0.0001, при изменении аргумента x
в указанном диапазоне [a, b] с шагом h.

Для сравнения в каждой точке вычислить также функцию  
y = f(x), являющуюся аналитическим выражением ряда.

---

1. s = Σ от i = 0 до ∞ [ (-1)^i * x^(2i) / (2i)! ]

   y = cos(x)

   a = 0.1  
   b = 1  
   h = 0.1


```python
import math as m
def summa(x):
    s=0
    for i in range(20):
        factorial = 1
        for j in range(1, 2*i+1):
            factorial = factorial*j
            
        s=s+(pow(-1, i))*pow(x, 2*i)/factorial
    return s

cos60 = summa(3.14159/3)
print(cos60, m.cos(3.14159/3))
```

7. В компьютер вводятся координаты и точек, лежащих на плоско- сти. После ввода координат каждой точки выводить номер

```python
count1 = 0
count3 = 0
i = 0

while True:
    try:
        i += 1
        x = float(input())
        y = float(input())
        
        if x > 0 and y > 0:
            quadrant = 1
            count1 += 1
        elif x < 0 and y > 0:
            quadrant = 2
        elif x < 0 and y < 0:
            quadrant = 3
            count3 += 1
        elif x > 0 and y < 0:
            quadrant = 4
        else:
            quadrant = 0
        
        print(f"Точка {i} находится в {quadrant}-м квадранте.")
    
    except:
        break 

print(f"Количество точек в 1-м квадранте: {count1}")
print(f"Количество точек в 3-м квадранте: {count3}")
```

5. Упорядочить по возрастанию элементы массива с четными ин- дексами (остальные элементы оставить на своих местах).

```python
n = int(input())
s = input().split()  

a = [] 
for i in range(n):
    a.append(int(s[i]))

even_idx = []
for i in range(0, n, 2):
    even_idx.append(a[i])

even_idx.sort()

j = 0
for i in range(0, n, 2):
    a[i] = even_idx[j]
    j += 1

for i in range(n):
    print(a[i], end=' ')
```

Лаба 4. Программу составить в двух вариантах, представляя матрицу: а) в виде двухмерного массива; б) в виде одномерной последовательности
1. В матрице размером 7 х 5 переставить строки таким образом, чтобы минимальные элементы строк следовали в порядке убывания.

```python
n_rows = 7
n_cols = 5

A = []
for i in range(n_rows):
    string_i = []  
    for j in range(n_cols):
        #value = float(input("Введи значение: "))
        value = round((random.random() - 0.5)*18, 2) 
        string_i.append(value)
    A.append(string_i)

mins = []
for i in range(n_rows):
    min_val = min(A[i])
    mins.append( (min_val, i) )

mins.sort(reverse=True) 

new_A = []
for min_val, old_i in mins:
    new_A.append(A[old_i])

print("Полученная матрица:")
for row in new_A:
    print(row)
```

```python
n_rows = 7
n_cols = 5
total = n_rows * n_cols

A = []
for i in range(n_rows):
    for j in range(n_cols):
        #value = float(input("Введи значение: "))
        value = round((random.random() - 0.5)*18, 2)
        A.append(value)

mins = []
for i in range(n_rows):
    start = i * n_cols
    row = A[start:start + n_cols]
    min_val = min(row)
    mins.append( (min_val, i) )

mins.sort(reverse=True)

new_A = []
for min_val, old_i in mins:
    start = old_i * n_cols
    for j in range(n_cols):
        new_A.append(A[start + j])

print("Полученная матрица:")
for i in range(n_rows):
    start = i * n_cols
    print(new_A[start:start + n_cols])
```

2. В заданной матрице расположить элементы четных строк в порядке возрастания, а элементы нечетных строк - в порядке убывания. Обработку матрицы по строкам осуществлять в методе. Для упорядочения строки использовать делегат.

```python
def process_row(row, row_index):
    if row_index % 2 == 0:  
        return sorted(row) 
    else: 
        return sorted(row, reverse=True)  

rows = 7
cols = 5
matrix = []
for i in range(rows):
    row = []
    for j in range(cols):
        value = round((random.random() - 0.5) * 18, 2)
        #value = float(input("Введите значение: ")) 
        row.append(value)
    matrix.append(row)

print("Исходная матрица:")
for row in matrix:
    print(row)


for i in range(rows):
    matrix[i] = process_row(matrix[i], i) 

print("После обработки:")
for row in matrix:
    print(row)
```

Лаба 6. 2. Соревнования по футболу между командами проводятся в два этапа. Для проведения первого этапа участники разбиваются на две группы по 12 команд. Для проведения второго этапа выбирается 6 лучших команд каждой группы по результатам первого этапа. Со- ставить список команд участников второго этапа.

```python
class team:
    def __init__(self, name, points):
        self.name = name
        self.points = points

def return_second_stage(group1, group2):
    result = []

    for group in [group1, group2]:
        indexes = list(range(len(group)))

        for i in range(len(group)):
            for j in range(0, len(group) - i - 1):
                if group[indexes[j]].points < group[indexes[j + 1]].points:
                    indexes[j], indexes[j + 1] = indexes[j + 1], indexes[j]

        sorted_group = [group[i] for i in indexes]

        for i in range(6):
            result.append(sorted_group[i].name)

    return result

group1 = [
    team("Team A1", 25), team("Team A2", 18), team("Team A3", 30),
    team("Team A4", 22), team("Team A5", 27), team("Team A6", 19),
    team("Team A7", 15), team("Team A8", 10), team("Team A9", 14),
    team("Team A10", 12), team("Team A11", 8), team("Team A12", 5)
]

group2 = [
    team("Team B1", 28), team("Team B2", 21), team("Team B3", 26),
    team("Team B4", 24), team("Team B5", 20), team("Team B6", 23),
    team("Team B7", 17), team("Team B8", 11), team("Team B9", 13),
    team("Team B10", 9), team("Team B11", 7), team("Team B12", 6)
]

second_stage = return_second_stage(group1, group2)

print("Команды второго этапа:")
for team_name in second_stage:
    print(team_name)
```

Лаба 6. 3. В соревнованиях участвуют три команды по 6 человек. Резуль- таты соревнований представлены в виде мест участников каждой ко- манды (1 - 18). Определить команду-победителя, вычислив количе- ство баллов, набранное каждой командой. Участнику, занявшему 1-е место, начисляется 5 баллов, 2-е - 4, 3-е - 3, 4-е 2,5-1, осталь- ным - 0 баллов. При равенстве баллов победителем считается коман- да, за которую выступает участник, занявший 1-е место.

```python
class team:
    def __init__(self, name, places):
        self.name = name
        self.places = places
        self.points = self.calculate_points()

    def calculate_points(self):
        score = 0
        for place in self.places:
            if place == 1:
                score += 5
            elif place == 2:
                score += 4
            elif place == 3:
                score += 3
            elif place == 4:
                score += 2
            elif place == 5:
                score += 1
        return score

def return_table(list_of_teams):
    table = []

    indexes = list(range(len(list_of_teams)))

    for i in range(len(list_of_teams)):
        for j in range(0, len(list_of_teams) - i - 1):
            t1 = list_of_teams[indexes[j]]
            t2 = list_of_teams[indexes[j + 1]]
            if t1.points < t2.points:
                indexes[j], indexes[j + 1] = indexes[j + 1], indexes[j]
            elif t1.points == t2.points:
                if min(t1.places) > min(t2.places):
                    indexes[j], indexes[j + 1] = indexes[j + 1], indexes[j]
    sorted_teams = [list_of_teams[i] for i in indexes]
    for i, team_ in enumerate(sorted_teams):
        row = []
        row.append(i + 1)
        row.append(team_.name)
        row.append(team_.points)
        table.append(row)

    return table
```

```python
t1 = team("Команда A", [1, 6, 7, 10, 12, 15])
t2 = team("Команда B", [2, 3, 8, 9, 14, 16])
t3 = team("Команда C", [4, 5, 11, 13, 17, 18])

teams = [t1, t2, t3]

result = return_table(teams)

print("Место | Команда | Баллы")
for row in result:
    print(row)
```
