2. Составить программу для обработки результатов кросса на 500 м для женщин. Для каждой участницы ввести фамилию, группу, фамилию преподавателя, результат. Получить результирующую таблицу, упорядоченную по результатам, в которой содержится также информация о выполнении норматива. Определить суммарное коли-
чество участниц, выполнивших норматив.

```python
class cross_500m_woman:
    def __init__(self, surname, group, prep_surname, result):
        self.surname = surname
        self.group = group
        self.prep_surname = prep_surname
        self.result = result

def return_table(list_of_exemplars, normativ):
    list_of_womans = []
    number_of_passed = 0
    indexes = list(range(len(list_of_exemplars)))

    for i in range(len(list_of_exemplars)):
        for j in range(0, len(list_of_exemplars)-i-1):
            if list_of_exemplars[indexes[j]].result > list_of_exemplars[indexes[j+1]].result:
                indexes[j], indexes[j+1] = indexes[j+1], indexes[j]

    sorted_exemplars = [list_of_exemplars[i] for i in indexes]
    for woman in sorted_exemplars:
        current_woman = []
        current_woman.append(woman.surname)
        current_woman.append(woman.group)
        current_woman.append(woman.prep_surname)
        current_woman.append(woman.result)
        if woman.result <= normativ: 
            current_woman.append('passed')
            number_of_passed = number_of_passed+1
        else: current_woman.append('unpassed')
        list_of_womans.append(current_woman)
    return list_of_womans, number_of_passed
```

```python
import random

surnames = ["Ivanova", "Petrova", "Sidorova", "Smirnova", "Volkova", 
            "Morozova", "Kuznetsova", "Popova", "Fedorova", "Orlova"]
groups = ["1", "2"]
prep_surnames = ["Nikolaev", "Fomin", "Voronov", "Pavlov", "Lebedev"]

list_of_exemplars = []

for _ in range(30):
    surname = random.choice(surnames)
    group = random.choice(groups)
    prep = random.choice(prep_surnames)
    result = round(random.uniform(90, 180), 2) 
    list_of_exemplars.append(cross_500m_woman(surname, group, prep, result))

normativ = 120 
table, passed_count = return_table(list_of_exemplars, normativ)

for row in table:
    print(row)
print(f"Количество участниц, выполнивших норматив: {passed_count}")
```

5. Соревнования по прыжкам на лыжах со 120-метрового трам- плина оценивают 5 судей. Каждый судья выставляет оценку за стиль прыжка по 20-балльной шкале. Меньшая и большая оценки отбрасы- ваются, остальные суммируются. К этой сумме прибавляются очки за дальность прыжка: 120 м 60 очков, за каждый метр превышения добавляются по 2 очка, при меньшей дальности отнимаются 2 очка за каждый метр. Получить итоговую таблицу соревнований, содержа- щую фамилию и итоговый результат для каждого участника в поряд-
ке занятых мест.

```python
class skijump:
    def __init__(self, surname, s1, s2, s3, s4, s5, meters):
        self.surname = surname
        self.score = ((s1+s2+s3+s4+s5) - max([s1,s2,s3,s4,s5]) - min([s1,s2,s3,s4,s5]))/3 + 60 + 2*(meters-120)

def return_table(list_of_skijumps):
    jumps = []

    indexes = list(range(len(list_of_skijumps)))

    for i in range(len(list_of_skijumps)):
        for j in range(0, len(list_of_skijumps)-i-1):
            if list_of_skijumps[indexes[j]].score < list_of_skijumps[indexes[j+1]].score:
                indexes[j], indexes[j+1] = indexes[j+1], indexes[j]
    sorted_skijumps = [list_of_skijumps[i] for i in indexes]
    
    for i, jump in enumerate(sorted_skijumps):
        jumper = []
        jumper.append(i+1)
        jumper.append(jump.surname)
        jumper.append(jump.score)
        jumps.append(jumper)

    return jumps
```

```python
jumps = [
    skijump("Ivanov", 18.5, 19.0, 18.0, 19.5, 18.5, 125),
    skijump("Petrov", 17.5, 18.0, 18.0, 17.0, 18.5, 122),
    skijump("Sidorov", 19.0, 19.5, 19.0, 18.5, 19.0, 130),
    skijump("Kuznetsov", 16.5, 17.0, 16.0, 17.5, 16.5, 118),
]

table = return_table(jumps)

for row in table:
    print(row)
```

1. Результаты сессии содержат оценки 5 экзаменов по каждой группе. Определить средний балл для трех групп студентов одного потока и выдать список групп в порядке убывания среднего балла. Ре- зультаты вывести в виде таблицы с заголовком.

```python
class group:
    def __init__(self, name, e1, e2, e3, e4, e5):
        self.name = name
        self.avg = (e1 + e2 + e3 + e4 + e5) / 5


def return_table(list_of_groups):
    table = []
    indexes = list(range(len(list_of_groups)))
    for i in range(len(list_of_groups)):
        for j in range(0, len(list_of_groups) - i - 1):
            if list_of_groups[indexes[j]].avg < list_of_groups[indexes[j + 1]].avg:
                indexes[j], indexes[j + 1] = indexes[j + 1], indexes[j]
    sorted_groups = [list_of_groups[i] for i in indexes]
    for i, grp in enumerate(sorted_groups):
        row = []
        row.append(i + 1)
        row.append(grp.name)
        row.append(grp.avg)
        table.append(row)
    return table
```

```python
g1 = group("Группа А", 4, 5, 3, 4, 5)
g2 = group("Группа Б", 3, 4, 4, 3, 4)
g3 = group("Группа В", 5, 5, 4, 5, 5)

groups = [g1, g2, g3]

result = return_table(groups)

print("Место | Группа | Средний балл")
for row in result:
    print(row)
```
