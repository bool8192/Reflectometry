16. В каждой строке матрицы А размером n х m максимальный элемент поместить в конец строки, сохранив порядок остальных элeментов.

```python
#m = int(input("Введите m: "))
m = 5
#n = int(input("Введите n: "))
n = 3
A = []

for i in range(n):
    string_i = []
    for j in range(m):
        string_i.append(round((random.random()-0.5)*18,2))
        #value = input("Введите значение: ")
        #string_i.append(value)
    A.append(string_i)
print("исходная матрица: ", A)


for i in range(n):
    max_value_in_string_i = float('-inf')
    for j in range(m):
        if A[i][j] > max_value_in_string_i: max_value_in_string_i = A[i][j]
    for j in range(m):
        if A[i][j] == max_value_in_string_i: idx_max_value_in_string_i = j
    if idx_max_value_in_string_i != m:
        A[i] = A[i][:idx_max_value_in_string_i]+A[i][idx_max_value_in_string_i+1:]+[(A[i][idx_max_value_in_string_i])]

print("Полученная матрица: ", A)
```

5. Дана матрица А размером 7 х 5. Максимальный элемент столбца заменить полусуммой первого и последнего элементов столбца, если максимальный элемент меньше этой полусуммы, в противном случае максимальный элемент заменить его номером в столбце.

```python
#m = int(input("Введите m: "))
m = 5
#n = int(input("Введите n: "))
n = 7
A = []

for i in range(n):
    string_i = []
    for j in range(m):
        string_i.append(round((random.random()-0.5)*18,2))
        #value = input("Введите значение: ")
        #string_i.append(value)
    A.append(string_i)
print("исходная матрица: ", A)


for i in range(n):
    max_value_in_string_i = float('-inf')
    for j in range(m):
        if A[i][j] > max_value_in_string_i: max_value_in_string_i = A[i][j]
    for j in range(m):
        if A[i][j] == max_value_in_string_i: idx_max_value_in_string_i = j
    if A[i][idx_max_value_in_string_i] < (A[i][0] + A[i][m-1])/2: A[i][idx_max_value_in_string_i] = (A[i][0] + A[i][m-1])/2
    else: A[i][idx_max_value_in_string_i] = idx_max_value_in_string_i+1 #в задании сказано - номер, а не индекс.

print("Полученная матрица: ", A)
```

2. Заполнить нулями элементы квадратной матрицы, расположенные по ее периметру (использовать один цикл).

```python
n = 7
A = []

for i in range(n):
    string_i = []
    for j in range(n):
        string_i.append(round((random.random()-0.5)*18,2))
        #value = input("Введите значение: ")
        #string_i.append(value)
    A.append(string_i)
#print(A)

p = 4*(n-1)
x = list(range(n)) + (n-1)*[n-1] + list(range(n-2, -1, -1)) + (n-2)*[0]
y = (n-1)*[0] + list(range(n)) + (n-1)*[n-1] + list(range(n-2, 0, -1))

for i in range(p):
    A[x[i]][y[i]] = 0
print("Полученная матрица: ", A)
```

```python
n = 3 

A = []
for i in range(n):
    for j in range(n):
        value = round((random.random()-0.5)*18,2)
        #value = float(input("Введите значение: ")) 
        A.append(value)


perimeter_indices = (
    list(range(n)) +                          
    [i * n + (n-1) for i in range(1, n-1)] +  
    list(range((n-1)*n, n*n)) +              
    [i * n for i in range(1, n-1)]          
)

# и наконец ОДИН цикл — хуярим нули
for idx in perimeter_indices:
    A[idx] = 0

print("Полученная матрица:", A)
```
