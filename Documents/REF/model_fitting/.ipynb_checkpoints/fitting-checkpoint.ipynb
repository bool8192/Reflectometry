```python
def objective(trial):

    lr1 = trial.suggest_float('lr1', 0.001, 0.1)
    lr2 = trial.suggest_float('lr2', 0.001, 0.1)
    lr3 = trial.suggest_float('lr3', 0.0002, 0.05)
    lr4 = trial.suggest_float('lr4', 0.0001, 0.04)
    lr5 = trial.suggest_float('lr5', 0.0001, 0.03)
    #lr6 = trial.suggest_float('lr6', 0.0001, 0.02)

    # Вызываем нашу функцию с новыми параметрами
    ans_d, ans_r_rho, ans_i_rho, ans_I, ans_Ibkg, loss, rel_losses = pie_optimizer(
        q, r, loss_function,
        betas=(0.9655, 0.99), gamma=0.988, wd=0.004,
        learning_rates=[lr1, lr2, lr3, lr4, lr5],
        iterations=iterations,
        loss_classes=loss_classes
    )

    return loss
```

```python
import matplotlib.pyplot as plt

# параметры
N = 200
device = "cpu"  # можно поменять на "cuda"
k = torch.arange(0, N, device=device)

# значения alpha2, которые хотим посмотреть
alpha_vals = [-0.5, -1.0, -2.0]
rough_vals = [-0.25, -1.0, -4.0]

def f(k, alpha2, rough2, N):
    return torch.exp((rough2)*torch.exp((-3+(7.5*k)/N)*alpha2))
    #-1*torch.exp(-1*torch.exp((-3+((-k+N)/N)*7.5))*alpha2)

plt.figure(figsize=(8,5))

for b in rough_vals:
    for a in alpha_vals:
        y = f(k, a, b, N).cpu()
        plt.plot(k.numpy(), y.numpy(), label=f"alpha2={a}")

plt.xlabel("k")
plt.ylabel("f(k)")
plt.title("График f(k, alpha2) при разных alpha2")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
from scipy.special import betainc
import matplotlib.pyplot as plt

def beta_cdf_transition(x, x0, L, rho1, rho2, B, U0, eta):
    """
    Переходный слой между rho1 и rho2 по Beta CDF.

    Параметры:
        x     : тензор координат (numpy array)
        x0    : начало переходного слоя (float)
        L     : длина переходного слоя (float, >0)
        rho1  : значение слева
        rho2  : значение справа
        alpha : параметр формы Beta (>0)
        beta  : параметр формы Beta (>0)

    Возвращает:
        rho(x): массив той же формы, что x
    """
    # Нормализованная координата
    u = (x - x0) / L

    # Ограничение значений в диапазоне [0, 1]
    u_clamped = torch.clamp(u, 0.0, 1.0)

    # Регуляризированная неполная бета-функция (Beta CDF)
    F = torch.pow((1+torch.exp(-B*(u - U0))), -1/eta)

    return rho1 + (rho2 - rho1) * F


# Генерация точек для графика
x = torch.linspace(0, 10, 500)

# Вычисление значения функции перехода
y1 = beta_cdf_transition(x, x0=0.0, L=10.0, rho1=0.0, rho2=1.0, B=10.0, U0= 0.5, eta=0.1)
y2 = beta_cdf_transition(x, x0=0.0, L=10.0, rho1=0.0, rho2=1.0, B=10.0, U0= 0.5, eta=10.1)
y3 = beta_cdf_transition(x, x0=0.0, L=10.0, rho1=0.0, rho2=1.0, B=10.0, U0= 0.2, eta=1.0)
y4 = beta_cdf_transition(x, x0=0.0, L=10.0, rho1=0.0, rho2=1.0, B=10.0, U0= 0.8, eta=1.0)

# Отображение графика
plt.figure(figsize=(8, 5))
plt.plot(x, y1, x, y2, x, y3, x, y4)
plt.xlabel('x')
plt.ylabel(r'$\rho(x)$')
plt.title('Переход по распределению Бета (CDF)')
plt.grid(True)
plt.show()
```

```python
import torch
import math

def bernstein_transition(x, L, rho1, rho2, coeffs):
    """
    Переходный слой через полином Бернштейна.

    x      : тензор координат
    L      : длина переходного слоя (>0)
    rho1   : значение слева
    rho2   : значение справа
    coeffs : список/тензор коэффициентов [c0,...,cn], где c0=0, cn=1 и c неубывающие

    Возвращает rho(x).
    """
    u = x / L
    u = torch.clamp(u, 0.0, 1.0)

    n = len(coeffs) - 1
    coeffs = torch.as_tensor(coeffs, dtype=u.dtype, device=u.device)

    # базис Бернштейна
    F = torch.zeros_like(u)
    for k in range(n+1):
        binom = math.comb(n, k)
        F = F + coeffs[k] * binom * (u**k) * ((1-u)**(n-k))

    return rho1 + (rho2 - rho1) * F

import matplotlib.pyplot as plt

x = torch.linspace(0, 10, 8)

# Степень 4 (n=4), c0=0, c4=1, внутри "свободные" параметры
y = []
for a in [0.01, 0.3, 0.4, 0.5, 0.7, 0.99]:
    coeffs = [0.0, 0.1*a, 0.5*a, 0.52*a, 0.7*a, 0.71*a, 0.9*a,  1.0]
    y.append(bernstein_transition(x, L=10.0, rho1=0.0, rho2=10.0, coeffs=coeffs))

for i, yi in enumerate(y):
    plt.plot(x.numpy(), yi.numpy(), label=f'a={i+1}')

# Оформляем график
plt.legend()
plt.title("Bernstein transition (n="+str(len(y))+")")
plt.grid(True)
plt.show()
```
