```python
# Install the assemblyai package by executing the command "pip install assemblyai"

import assemblyai as aai
import os

aai.settings.api_key = "436f8bb19e564ee7ace099a326d80c2f"

audio_file = "C:/Users/ya/Downloads/ffmpeg-2025-08-07-git-fa458c7243-full_build/bin/output.mp3"
#audio_file = "https://assembly.ai/wildfires.mp3"

# Конфигурация для русского языка
config = aai.TranscriptionConfig(
    language_code="ru",  # Явно указываем русский язык
    speech_model=aai.SpeechModel.best,
    punctuate=True,     # Автоматическая расстановка пунктуации
    format_text=True    # Форматирование текста (абзацы, заглавные буквы)
)

# Создаем транскрайбер
transcriber = aai.Transcriber()

# Транскрибируем аудио - передаем аргументы позиционно
try:
    transcript = transcriber.transcribe(audio_file, config)
except Exception as e:
    raise RuntimeError(f"Transcription failed: {str(e)}")

# Проверяем статус
if transcript.status == aai.TranscriptStatus.error:
    raise RuntimeError(f"Transcription failed: {transcript.error}")

# Получаем распознанный текст
russian_text = transcript.text
print("Распознанный текст:")
print(russian_text)

# Формируем путь для текстового файла (в той же директории)
text_file_path = os.path.splitext(audio_file)[0] + ".txt"

# Сохраняем в текстовый файл
try:
    with open(text_file_path, "w", encoding="utf-8") as f:
        f.write(russian_text)
    print(f"\nТекст успешно сохранен в файл:\n{text_file_path}")
except Exception as e:
    print(f"\nОшибка при сохранении файла: {str(e)}")
```
