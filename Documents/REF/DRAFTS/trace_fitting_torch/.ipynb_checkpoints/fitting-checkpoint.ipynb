```python
class Comparator:
    def __init__(self, q, orig_matr):
        self.q = q
        self.original_trace = reflectometry_trace(q, orig_matr)
        
    def compare_trace(self, var_matr):
        crit1, r_max1, q_max1, r_min1, q_min1 = self.original_trace 
        crit2, r_max2, q_max2, r_min2, q_min2 = reflectometry_trace(self.q, var_matr)
        #print(var_matr)
        
        loss = (2*(crit1-crit2)/crit1)**2 + torch.sum(((r_max1-r_max2)/r_max1)**2) + torch.sum(((q_max1-q_max2)/q_max1)**2) + torch.sum(((q_min1-q_min2)/q_min1)**2)
        
        return loss

# Вызываем compare много раз без повторной загрузки reference_points
loss_function = Comparator(q, matr)
```

```python
class varbounds:
    def __init__(self, bounds_tensor):
        self.bounds = bounds_tensor.clone()

    def objective_function(self, var_vector):
        matrixx = self.bounds[..., 0].clone()
        matrixx[torch.where(self.bounds[..., 0] != self.bounds[..., 1])] = var_vector.clone()
        return loss_function.compare_trace(matrixx)

objective_function = varbounds(all_bounds)
```

```python
def numerical_gradient(fun, x, eps=1e-10):
    x = x.detach().clone()  # Отключаем автоград и делаем независимую копию
    grad = torch.zeros_like(x)
    
    # Сохраняем исходную форму
    original_shape = x.shape
    x_flat = x.view(-1)
    grad_flat = grad.view(-1)
    
    for i in range(x_flat.numel()):
        # Создаем копии для каждого направления
        x_pos = x_flat.clone()
        x_neg = x_flat.clone()
        
        # Модифицируем конкретный элемент
        x_pos[i] = x_flat[i] + eps
        x_neg[i] = x_flat[i] - eps
        
        # Возвращаем исходную форму для вычисления функции
        f_pos = fun(x_pos.view(original_shape))
        f_neg = fun(x_neg.view(original_shape))
        
        # Вычисляем градиент
        diff = f_pos - f_neg
        grad_flat[i] = diff / (2 * eps)

    return grad.view(original_shape)
```

```python
def torch_minimize(fun, x0, lr=2e-18, max_iter=16000, tol=2e-23):
    x = x0.clone().detach()
    prev_loss = None

    for i in range(max_iter):
        loss = fun(x)
        grad = numerical_gradient(fun, x)
        #print(grad)
        
        x = x - lr * grad  # Шаг градиентного спуска
        
        if prev_loss is not None and abs(prev_loss - loss.item()) < tol:
            print(f"Converged at iteration {i}", "  prev", prev_loss, "   item", loss.item())
            break
        prev_loss = loss.item()
    return x, prev_loss
```

```python
def adamw(fun, x0, lr=3.2e-11, max_iter=16000, tol=2e-12):
    x = x0.clone().detach().requires_grad_(True)
    optimizer = torch.optim.AdamW([x], lr=lr)
    
    prev_loss = None
    
    for i in range(max_iter):
        optimizer.zero_grad()
        
        loss = fun(x)
        grad = numerical_gradient(fun, x.detach())
        x.grad = grad
        optimizer.step()
        if i%400 == 0: print(loss, "   ",grad,"  ",  x,  "\n")
        
        loss_val = loss.item()
        if prev_loss is not None and abs(prev_loss - loss_val) < tol and i>=600:
            print(f"Converged at iteration {i}, loss: {loss_val}")
            break
        prev_loss = loss_val
    
    return x.detach(), prev_loss
```
