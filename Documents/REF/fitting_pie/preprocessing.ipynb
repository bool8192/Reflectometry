```python
import numpy as np
import torch
import matplotlib.pyplot as plt

data = np.loadtxt('experiment.txt')
```

```python
o = data[:, 2]/data[:, 1]  
x = data[:, 0]

fig, ax = plt.subplots(figsize=(18, 8))
r = data[:, 1]
r2 = data[:, 1] + data[:, 2]
r1 = data[:, 1] - data[:, 2]
ax.fill_between(data[:, 0], r1, r2, color='purple', alpha=0.7, label= '1')
ax.set_yscale('log')

ax.legend()
plt.show()
```

```python
import math 
rn = r[r > 0.9*r1[0]][0:1+math.ceil(len(r[r > 0.9*r1[0]])*0.7)]
I0_plus = np.mean(rn)
```
